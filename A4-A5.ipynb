# Bus Reliability in Seattle

### Motivation & problem:
In Seattle, hundreds of thousands of people rely on the King County Metro bus system to get to work, school, and for other everyday commutes. When buses arrive late, are completely canceled, or skip stops, it obviously makes it harder for riders to be where they need to be on-time, which is not what we want. 

For my project, my overall goal is to investigate bus reliability across Seattle and explore how factors like the presence of dedicated bus lines impact on-time performance. Bus lines are designed to help buses avoid traffic, so do routes that have greater access to bus lanes experience less delays? I hope to identify routes with the most reliability issues and explore where potential improvements in transit infrastructure, particularly bus lines, could help.

### Data:
I plan to use the City of Seattle’s [Transit Lanes](https://data-seattlecitygis.opendata.arcgis.com/datasets/SeattleCityGIS::transit-lanes/about) dataset which provides information on where bus lines are located throughout Seattle, how long they 
run for, their operating hours, and any restrictions they may have. This dataset will be useful towards understanding how bus lanes are distributed throughout the city and I will use the GeoJSON version of the dataset to access the geographic coordinates of these bus lane segments.

The [King County Metro Routes](https://gis-kingcounty.opendata.arcgis.com/datasets/kingcounty::king-county-metro-routes/about) dataset provides geographic data on each bus route in King County, including their route paths and overall length. Again, I will use the GeoJSON version of the dataset to access the coordinates that distinguish each route and see how much they overlap with the locations/coordinates of bus lanes.

I also plan to use data from King County’s [Rider Dashboard](https://data.kingcounty.gov/Transportation-Roads/Metro-Rider-Dashboard/v229-zr3q/about_data) which provides data on on-time performance, ridership impact, missed stops, and bus capacity. The data can additionally be filtered by route, direction, month, time of day, and day type. I will mostly be looking at the on-time performance data, so I can compare reliability between different routes. 

All of these sources are available to the public through the City of Seattle and King County Metro. The Transit Lanes dataset and Ridership Dashboard have public domain licenses. King County Metro Routes terms of service states users are free to copy, distribute, and use the information, with the exception that the data is not to be sold.

## A5 Final Project Plan
### **Overview & research questions / hypotheses:**
For this project, I aim to investigate how on-time bus performance varies across different bus routes in Seattle and how access to dedicated bus lanes might play a factor in on-time performance.

**Main research question:**
- How does on-time performance vary across different bus routes in Seattle, and to what extent might access to dedicated bus lanes influence on-time performance?

**Additional research questions:**
- Which routes have the best and worst on-time performance?
- Which routes have the most and least access to bus lanes (% of route covered by bus lane)?
- Do routes with greater access to bus lanes have better on-time performance?

I expect the routes that run in the Downtown Seattle area experience worst on-time performance since this is typically the area in Seattle with the most congestion/traffic. I also expect routes with greater access to bus lanes to have better on-time performance. However, I do recognize that bus lanes are typically placed in areas with the most congestion, so just because a route has better bus lane coverage may not reflect in on-time performance.

### **Background / related work:**
We already know the purpose and intention of bus lanes is to help buses avoid traffic and improve transit times, especially when traffic is highly congested. According to the [Seattle Department of Transportation (SDOT)](https://www.seattle.gov/transportation/projects-and-programs/programs/transit-program/transit-lanes), they identify opportunities for bus lanes based on factors such as passenger demand, passenger delay, and travel time variability and reliability. Seattle Transit Blog's article on [Seattle Transit Lanes in 2025](https://seattletransitblog.com/2025/12/26/seattles-transit-lanes-in-2025/) also reports that the city currently has 55 miles of transit lanes, with much of those lanes concentrated in major transit corridors and Downtown Seattle. Together, this helps explain why transit lanes are implemented and how they are distributed across Seattle. Since access to bus lanes varies by route, it makes it interesting to see whether routes with greater transit lane coverage actually experience better on-time performance.

Between January 2019 and November 2016, about 79% of bus arrivals have been on-time while about 7% were early and 15% were late (source: [King County Metro's Rider Dashboard](https://data.kingcounty.gov/Transportation-Roads/Metro-Rider-Dashboard/v229-zr3q/about_data)). This slightly falls short of King County Metro's goal of 80% on-time performance. These percentages give me a baseline for my study, but I want to investigate how performance varies across individual routes because the overall average may give a misleading impression of how reliable certain routes actually are.

### **Methodology:**
I will be using Python to help with the quantitative analyis of my datasets and compare on-time performance across different routes.

#### Comparing geographic data/coordinates
I will first use geographic data from the [King County Metro Routes](https://gis-kingcounty.opendata.arcgis.com/datasets/kingcounty::king-county-metro-routes/about) and [Transit Lanes](https://data-seattlecitygis.opendata.arcgis.com/datasets/SeattleCityGIS::transit-lanes/about) to compare bus route paths with dedicated bus lane locations. By comparing the paths (coordinates) of the two datasets, I can see what portions of bus routes overlap with transit lanes. I will then calculate the length of those portions and divide it by the total length of each route to get the percentage of which each bus route is covered by dedicated bus lanes. Doing this gives me a consistent way to compare bus lane coverage across different routes.

#### Analyzing on-time performance
I will collect the overall on-time performance percentages for each bus route from January 2025 to present from the [Rider Dashboard](https://data.kingcounty.gov/Transportation-Roads/Metro-Rider-Dashboard/v229-zr3q/about_data). I will include both directions, all time periods, and all days of the week so I can focus on overall on-time performance. I plan to use descriptive stats (mean, median, range) to summarize on-time performance across different routes to make it easy to compare how reliability varies between routes.

#### Correlational analysis between bus lane coverage and on-time performance
I will analyze the relationship between bus lane coverage and on-time performance for each route by doing correlational analysis. I will calculate the correlation coefficient (r) so I can better understand the strength of the relationship and see whether routes with greater bus lane coverage have better on-time performance. It is important to remember correlation is not causation as other factors such as traffic, route length, or ridership could affect on-time performance. It is also important to note that bus lanes are typically placed in the most congested areas.

#### Presenting my findings
I plan to present my findings through various data visualizations. Because my project contains geographic data, I think a heatmap according to each route's bus lane coverage would be helpful in showing routes with greater or lower access to bus lanes, and where bus lanes are most concentrated in Seattle. For my correlational analysis, I will create a scatterplot displaying the relationship between bus lane coverage and on-time performance, with each point representing a different bus route. To present my on-time performance findings, I plan to create a simple table with each route's on-time performance and estimated bus lane coverage, sorted from lowest to highest on-time percentage.

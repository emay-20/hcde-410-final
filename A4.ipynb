# Bus Reliability in Seattle

### Motivation & problem:
In Seattle, hundreds of thousands of people rely on the King County Metro bus system to get to work, school, and for other everyday commutes. When buses arrive late, are completely canceled, or skip stops, it obviously makes it harder for riders to be where they need to be on-time, which is not what we want. 

For my project, my overall goal is to investigate bus reliability across Seattle and explore how factors like the presence of dedicated bus lines impact on-time performance. Bus lines are designed to help buses avoid traffic, so do routes that have greater access to bus lanes experience less delays? I hope to identify routes with the most reliability issues and explore where potential improvements in transit infrastructure, particularly bus lines, could help.


### Data:
I plan to use the City of Seattle’s [Transit Lanes](https://data-seattlecitygis.opendata.arcgis.com/datasets/SeattleCityGIS::transit-lanes/about) dataset which provides information on where bus lines are located throughout Seattle, how long they 
run for, their operating hours, and any restrictions they may have. This dataset will be useful towards understanding how bus lanes are distributed throughout the city and, when paired with the King County Metro Stops dataset below, which bus routes have greater access to these dedicated lanes.

The [King County Metro Stops](https://gis-kingcounty.opendata.arcgis.com/datasets/kingcounty::king-county-metro-stops/about) dataset provides information on the bus stops throughout Seattle, including the routes that serve each stop and what streets they are located on. Again, when paired with the Transit Lines dataset, this information will help me understand which bus routes have access to dedicated bus lanes. It will also allow me to see which routes experience more reliability issues and whether those problems are concentrated in certain locations/streets.

I also plan to use data from King County’s [Rider Dashboard](https://data.kingcounty.gov/Transportation-Roads/Metro-Rider-Dashboard/v229-zr3q/about_data) which provides data on on-time performance, ridership impact, missed stops, and bus capacity. The data can additionally be filtered by route, direction, month, time of day, and day type. I will mostly be looking at the on-time performance data, so I can compare reliability between different routes. 

All of these sources are available to the public through the City of Seattle and King County Metro. The Transit Lanes dataset and Ridership Dashboard have public domain licenses. King County Metro stops terms of service states users are free to copy, distribute, and use the information, with the exception that the data is not to be sold.


### Unknowns & dependencies:
I think one major unknown is just whether the available datasets include missing or inaccurate information. Especially for the on-time performance data in the Ridership Dashboard, I have no way to verify whether the data reflects actual bus time arrivals. And since my project centers heavily around bus reliability/on-time performance, it could influence the overall accuracy of my analysis.
